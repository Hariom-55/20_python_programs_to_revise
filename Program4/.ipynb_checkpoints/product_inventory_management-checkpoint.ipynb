Problem Statement


Build a Product Inventory & Sales Management System that stores products, validates their information, tracks inventory, calculates sales statistics, identifies low-stock products, searches products, and generates a final analytical report.

The program should maintain product information using a dictionary.

Each product should contain:

Product ID
Product Name
Category
Price
Stock Quantity
Units Sold

```python
import sys
```

```python
#taking the input in a dictionary 

products = {}

while True:
    try:
        number_of_products = int(input("Enter the number of Products in the Inventory"))

        if number_of_products > 0 :
            break
        else :
            print("Number of Negative Should be a psoitive Integral Value")
            continue

    except ValueError :
        print("Please Enter a Integer Value")
        continue 
        



for product in range(number_of_products) : 

    product_name = input("Name of the Product: ")
    product_name = " ".join(product_name.strip().title().split())

    product_category = input("What is the Product Category: ")
    product_category = " ".join(product_category.strip().title().split()) 

    while True :

        try:
            price = float(input("Price of the Product: "))
    
            if price > 0 :
                break
            else :
                print("Price of the product cannot be Zero or negative")
                continue

        except ValueError :
            print("Invalid Input Type")
            continue 

    while True :

        try :
            stock_quantity = int(input("Total Stock of the Product: "))

            if stock_quantity > 0 :
                break
            else :
                print("Stock Quantity of the product cannot be Zero or negative")
                continue

        except ValueError :
            print("Invalid Input Type")
            continue 


    while True :

        try :
            units_sold = int(input("Total units sold: "))

            if 0 <= units_sold <= stock_quantity  :
                break
            elif units_sold > stock_quantity :
                print("Product sold cannot be more than the product quantity")
                continue 
            else :
                print("Stock Quantity of the product cannot be Zero or negative")
                continue

        except ValueError :
            print("Invalid Input Type")
            continue 


    product_ID = f"P{product+1:03d}"

    products[product_ID] = {
        "Name" : product_name ,
        "Category" : product_category ,
        "Price" : price ,
        "Stock Quantity" : stock_quantity,
        "Units Sold" : units_sold
    }
```

```python
#Inventory Calculation 

if len(products) > 0 :
    
    total_number_of_products = len(products) 
    total_inventory = 0 ;
    total_units_Sold = 0 ;
    total_inventory_value = 0 ;
    total_price_of_all_products = 0 ;
    total_sales_revenue = 0 ;
    highest_product_price = products[list(products.keys())[0]]["Price"]
    lowest_product_price = products[list(products.keys())[0]]["Price"] 


    for product_id , details in products.items() :

        total_inventory += details["Stock Quantity"]
        total_units_Sold +=  details["Units Sold"]
        total_inventory_value += details["Stock Quantity"] * details ["Price"]
        total_price_of_all_products += details["Price"]
        total_sales_revenue += details["Units Sold"] * details["Price"] 

        if details["Price"] > highest_product_price : 
            highest_product_price = details["Price"] 

        if details["Price"] < lowest_product_price : 
            lowest_product_price = details["Price"] 


    average_price_of_product = total_price_of_all_products / total_number_of_products 

    highest_priced_ProductID = [product_id 
                                for product_id , details in products.items() 
                                if details["Price"] == highest_product_price ] 
    
    lowest_priced_ProductID = [product_id 
                               for product_id , details in products.items()
                               if details["Price"] == lowest_product_price ]

else :
    print("No Product in the Inventory")
    sys.exit() 
    
```

```python
#Stock Classification 

high_stock_products = []
medium_stock_products = []
low_stock_products = []
out_of_stock_products = [] 

for product_id , details in products.items() : 
    remaining_stock = details["Stock Quantity"] - details["Units Sold"]

    if remaining_stock >= 50 :
        high_stock_products.append(product_id)

    elif remaining_stock >= 20 :
        medium_stock_products.append(product_id)

    elif remaining_stock >= 1 :
        low_stock_products.append(product_id)

    else :
        out_of_stock_products.append(product_id) 



        
```

```python
#Product Search 
while True :
    product_ID = input ("Enter the Product Id to search : ") 
    product_ID = product_ID.strip().upper() 
    
    if product_ID in products :
        
        details = products[product_ID] 
        remaining_stock = details["Stock Quantity"] - details["Units Sold"]
        
        print(f"""  PRODUCT DETAILS 
        Name: {details["Name"]}
        Category: {details["Category"]}
        Price: ${details["Price"]:.2f}
        Stock Quantity: {details["Stock Quantity"]}
        Units Sold: {details["Units Sold"]}
        Remaining Stock: {remaining_stock}
        """)
        break 
        
    else : 
        print("Product not Found")
        

        
```

```python
#ranking the Products 

ranked_products = [product_id for product_id in products] 

for i in range(len(ranked_products)) :
    for j in range(i+1 , len(ranked_products)) :
        if products[ranked_products[i]]["Units Sold"] * products[ranked_products[i]]["Price"] < products[ranked_products[j]]["Units Sold"] * products[ranked_products[j]]["Price"] :
            ranked_products[i] , ranked_products[j] = ranked_products[j] , ranked_products[i]
```

```python
#category Analysis 

category_analysis = {} 

for product_id , details in products.items() :
    category = details["Category"] 

    if category not in category_analysis :
        category_analysis[category] = {
            "Total Products" : 0 ,
            "Total Units Sold" : 0 ,
            "Total Sales Revenue" : 0 ,
        }

    category_analysis[category]["Total Products"] += 1
    category_analysis[category]["Total Units Sold"] += details["Units Sold"]    
    category_analysis[category]["Total Sales Revenue"] += details["Units Sold"] * details["Price"]       
```

```python
#printing the report

print("================ PRODUCT ANALYTICAL REPORT =====================", end="\n\n")

print("Total Number of Products: ", total_number_of_products)
print("Total Inventory: ", total_inventory)
print("Total Units Sold: ", total_units_Sold)
print()

print("Total Inventory Value: ", total_inventory_value)
print("Total Sales Revenue: ", total_sales_revenue)
print("Average Product Price: ", average_price_of_product)
print()


print("================ HIGHEST / LOWEST PRICED PRODUCTS ================", end="\n\n")

print("Highest Product Price: ", highest_product_price)

print("Highest Priced Product(s):")
for product_id in highest_priced_ProductID:
    print(f"  Product ID: {product_id}")
    print(f"  Product Name: {products[product_id]['Name']}")
    print(f"  Price: ${products[product_id]['Price']:.2f}")
    print()

print("Lowest Product Price: ", lowest_product_price)

print("Lowest Priced Product(s):")
for product_id in lowest_priced_ProductID:
    print(f"  Product ID: {product_id}")
    print(f"  Product Name: {products[product_id]['Name']}")
    print(f"  Price: ${products[product_id]['Price']:.2f}")
    print()


print("================ STOCK DISTRIBUTION ================", end="\n\n")

print("High Stock: ", len(high_stock_products))
print("Medium Stock: ", len(medium_stock_products))
print("Low Stock: ", len(low_stock_products))
print("Out of Stock: ", len(out_of_stock_products))
print()


print("================ CATEGORY ANALYSIS ================", end="\n\n")

for category, stats in category_analysis.items():
    print(f"Category: {category}")
    print(f"  Total Products: {stats['Total Products']}")
    print(f"  Total Units Sold: {stats['Total Units Sold']}")
    print(f"  Total Sales Revenue: ${stats['Total Sales Revenue']:.2f}")
    print()


print("================ PRODUCT RANKING ================", end="\n\n")

for rank, product_id in enumerate(ranked_products, start=1):
    details = products[product_id]

    sales_revenue = details["Units Sold"] * details["Price"]

    print(
        f"Rank {rank}: "
        f"Product ID: {product_id}, "
        f"Name: {details['Name']}, "
        f"Units Sold: {details['Units Sold']}, "
        f"Sales Revenue: ${sales_revenue:.2f}"
    )

print()


print("================ LOW STOCK ALERT ================", end="\n\n")

if len(low_stock_products) > 0:

    for product_id in low_stock_products:

        details = products[product_id]

        remaining_stock = (
            details["Stock Quantity"] - details["Units Sold"]
        )

        print(f"Product ID: {product_id}")
        print(f"Product Name: {details['Name']}")
        print(f"Category: {details['Category']}")
        print(f"Remaining Stock: {remaining_stock}")
        print()

else:
    print("No Low Stock Products")
    print()


print("================ OUT OF STOCK PRODUCTS ================", end="\n\n")

if len(out_of_stock_products) > 0:

    for product_id in out_of_stock_products:

        details = products[product_id]

        print(f"Product ID: {product_id}")
        print(f"Product Name: {details['Name']}")
        print(f"Category: {details['Category']}")
        print()

else:
    print("No Out of Stock Products")
    print()


print("================ DUPLICATE PRODUCT NAMES ================", end="\n\n")

duplicate_product_names = []

product_ids = list(products.keys())

for i in range(len(product_ids)):
    for j in range(i + 1, len(product_ids)):

        product_i = product_ids[i]
        product_j = product_ids[j]

        if products[product_i]["Name"] == products[product_j]["Name"]:

            if products[product_i]["Name"] not in duplicate_product_names:
                duplicate_product_names.append(
                    products[product_i]["Name"]
                )

if len(duplicate_product_names) > 0:

    for name in duplicate_product_names:
        print("Duplicate Product Name:", name)

else:
    print("No Duplicate Product Names Found") 



```

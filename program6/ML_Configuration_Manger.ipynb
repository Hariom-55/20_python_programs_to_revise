```python
model_configuration = [] 

while True : 
    try:
        total_configuration = int(input("Enter the total number of ML configuration you want to store : "))

        if total_configuration > 0 :
            break 
        else :
            print("Invalid Input Value it cannot be less than Zero")
            continue 

    except ValueError :
        print("Invalid Input Type !!! ")
        continue 


for configuration in range(total_configuration) :
    
    while True :
        try :
            model_name = input("Enter the Model Name : ").strip() 

            if model_name == "" :
                print("Model Name cannot be Empty")
                raise ValueError
            else :
                model_name = " ".join(model_name.title().split()) 
                break

        except ValueError :
            continue 


    while True :
        try:
            hyperparameter_value = int(input("Enter the Value of parameter : "))
    
            if hyperparameter_value > 0 :
                break 
            else :
                print("Invalid Input")
                continue 

        except ValueError :
            print("Invalid Input Type !!!")
            continue 


    while True :
        try :
            random_state = int(input("Enter the random state : "))

            if random_state >= 0 :
                break
            else :
                print("Invlaid Input")
                continue 

        except ValueError :
            print("INvalid Input Type !!!")
            continue 


    model_configuration.append((model_name, hyperparameter_value, random_state))        
```

```python
#Configuration Analysis
import sys 
no_of_configuration = len(model_configuration)

if no_of_configuration > 0 : 
    #Number of Configuration of each model 

    model_count = {} #all the model and their configuration 
    highest_parameter_value = model_configuration[0][1] 
    lowest_parameter_value = model_configuration[0][1] 
    sum_parameter_value = 0 

    for model in model_configuration : 
        sum_parameter_value += model[1] 

        if model[0] in model_count :
            model_count[model[0]] +=1

        else :
            model_count[model[0]] = 1


        if model[1] > highest_parameter_value :
            highest_parameter_value = model[1] 

        if model[1] < lowest_parameter_value :
            lowest_parameter_value = model[1] 

    
else :
    sys.exit()

    
```

```python
#model_search operation 

name = input("Enter the Model Name to Search : ").strip().title().split() 

model_name_to_search = " ".join(name) 

if model_name_to_search in model_count :
    print("Model Exist in the Database") 

    for model in model_configuration :
        if model[0] == model_name_to_search:
            print(f"""
            Model Name : {model[0]}
            Hyperparameter Value : {model[1]}
            Random State : {model[2]}
            """)
            print()

else :
    print("Model does not exist") 
```

```python
#Configuration Comparison 

first_model = input("Enter the name of First Model : ").strip().title().split()
first_model = " ".join(first_model)

second_model = input("Enter the name of Second Model : ").strip().title().split()
second_model = " ".join(second_model) 

first_configuration = None 
second_configuration = None 

for configuration in model_configuration : 

    if configuration[0] == first_model and first_configuration is None :
        first_configuration = configuration 

    elif configuration[0] == second_model and second_configuration is None :
        second_configuration = configuration 

    

if (first_configuration is None or second_configuration is None) :
    print("Model name you entered may not exist")
    
elif first_configuration == second_configuration :
    print("Configuration are identical , Each of the configuration paraemeter as equal")

else :
    print("Configuration are different") 

    if first_configuration[0] == second_configuration[0] :
        print("Identical Names")
    else :
        print("Names are different")

    if first_configuration[1] == second_configuration[1] :
        print("Identical Hyperparameters")
    else :
        print("Hyperparameters are different")

    if first_configuration[2] == second_configuration[2] :
        print("Identical Random State")
    else :
        print("Random States are different")



    
```

```python
#Duplicate Configuration Detection 

duplicate_configurations = []

for i in range(len(model_configuration)):
    for j in range(i + 1, len(model_configuration)):
        if model_configuration[i] == model_configuration[j]:
            if model_configuration[i] not in duplicate_configurations:
                duplicate_configurations.append(model_configuration[i]) 
```

```python
#Configuration Ranking

ranked_configuration = model_configuration.copy() 

for i in range(len(ranked_configuration)):
    for j in range(i+1, len(ranked_configuration)) :
        if ranked_configuration[i][1] < ranked_configuration[j][1] :
            temp = ranked_configuration[i] 
            ranked_configuration[i] = ranked_configuration[j]
            ranked_configuration[j] = temp 

            
```

```python
print("================ ML CONFIGURATION REPORT ================" , end ="\n\n")

print("Total Configurations: " , total_configuration ) 
print()

print("================ MODEL DISTRIBUTION ================", end = "\n\n")

for model , count in model_count.items() :
    print(f"{model} : {count}")

print() 

print("================ CONFIGURATION STATISTICS ================" , end ="\n\n" )

print("Highest Parameter Value: " ,highest_parameter_value)
print("Lowest Parameter Value: " , lowest_parameter_value)
print("Average Parameter Value: ", sum_parameter_value / total_configuration )

print()

print("================ DUPLICATE CONFIGURATION ================" , end ="\n\n" )

if len(duplicate_configurations) > 0 :
    
    for configuration in duplicate_configurations :
        print(f"""
        Model Name : {configuration[0]}
        Hyperparameter Value : {configuration[1]}
        Random State : {configuration[2]}
        """)
        print() 

else :
    print("No Duplicate Configurations Found")

print()

print("================ CONFIGURATION RANKING ================" , end ="\n\n" )

for rank , configuration in enumerate(ranked_configuration , start = 1) :
    print(f"""
    Rank {rank}
    Model = {configuration[0]},
    Hyperparamter = {configuration[1]},
    Random State = {configuration[2]}
    """) 

    print()

```

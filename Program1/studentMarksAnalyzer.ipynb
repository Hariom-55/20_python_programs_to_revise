```python

import sys
students = int(input("Enter the total no. of students"))

average_of_students = [] 
subjects = ["Physics" , "Mathematics" , "English" , "Chemistry" , "Computer Science"]
for student in range(students):
    total_marks = 0
    for subject in subjects:
        
        while True:
            try:
                marks = float(input(f"Enter {subject} marks: "))
                if (marks < 0 or marks > 100) :
                    print("Invalid Marks Entered")
                    continue
    
                total_marks += marks
                break
            
            except ValueError: 
                print("Invalid Input")
                 
    calculated_average = total_marks/len(subjects)
    average_of_students.append(calculated_average) 


 # now we have average marks of all the students 

total_students = len(average_of_students) 

if total_students > 0:
    
    highest_average = average_of_students[0]
    lowest_average = average_of_students[0]
    total_passed_students = 0
    
    for average in average_of_students :
        if(average > 45):
            total_passed_students +=1
        if(average < lowest_average):
            lowest_average=average 
        elif(average > highest_average):
            highest_average=average
    
    percentage_of_passed_students = (total_passed_students/total_students)*100

    overall_average = sum(average_of_students)/total_students

    A = B = C = D = E = F = 0 # Grade of the students 
    
    for average in average_of_students : 
        if ( average >= 90 ):
            A+=1
        elif ( average >= 80 ):
            B+=1
        elif ( average >= 70 ):
            C+=1
        elif ( average >= 60 ):
            D+=1
        elif ( average >= 45 ):
            E+=1
        else:
            F+=1

    average_of_students.sort(reverse = True)
    
else :
    print("Number of students cannot be Zero")
    sys.exit()


print("=============== Student Marks Analyzer ===============")
print(f"Pass Percentage of the Examination: {percentage_of_passed_students}%")
print("Overall Average of the Examination: ",overall_average)
print("Highest Percentage Obtained in the Examination: ",highest_average)
print("Lowest Percentage Obtained  the Examination: ",lowest_average)
print("Grade Distribution")
print(f"""
A Grade: {A} 
B Grade: {B}
C Grade: {C} 
D Grade: {D}
E Grade: {E} 
F Grade: {F}

""")
print("Ranking of the Students")
print(average_of_students)  
```

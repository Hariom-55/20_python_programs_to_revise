
### Program 5 — Student Attendance Analyzer

The program should handle attendance for two classes/groups and use **sets** to perform comparisons.

Requirements:

1. Take the list of students present in **Class A**.
2. Take the list of students present in **Class B**.
3. Remove duplicate student entries.
4. Find students present in **both classes**.
5. Find students present only in Class A.
6. Find students present only in Class B.
7. Find the **union** of students from both classes.
8. Find students absent from a class when given the complete student list.
9. Calculate attendance statistics such as:

   * Total unique students
   * Number present in each class
   * Number common to both classes
   * Number absent
10. Display the final attendance analysis.




```python
#list of all students 

students = [] 

while True :
    try:
        total_students = int(input("Enter total number of students: ") ) 
    
        if total_students > 0 :
            break 
            
        else :
            print("Total number of Student should be greater than Zero !!!")
            continue 

    except ValueError :
        print("Invalid Input Type !!!")


#taking inputs 

for student in range(total_students) :

    student_name = input ("Enter the Name of Student : ") 
    student_name = " ".join(student_name.strip().title().split()) 

    students.append(student_name)



#creating both the classes and marking attendance  

class_a = []
class_b = [] 

print("Mark Attendance of Student in class A " ) 

while True :
    try :
        name_a = input ("Enter the Name of Present Student : ") 
        name_a = " ".join(name_a.strip().title().split())

        if name_a in students :
            class_a.append(name_a)

        else :
            print("Student not Registered")

    except EOFError :
        print("Attendance Marked for Class A!!")
        break ;



print("Mark Attendance of Student in class B " )

while True :
    try :
        name_b = input ("Enter the Name of Present Student : ") 
        name_b = " ".join(name_b.strip().title().split())

        if name_b in students :
            class_b.append(name_b)

        else :
            print("Student not Registered")


    except EOFError :
        print("Attendance Marked for Class B!!")
        break ;

```

```python
#implementing sets 

present_class_a = set(class_a) #creating Sets automatically removes duplicates  if any exists 
present_class_b = set(class_b)

#set operations :
# 1.Intersection
 common_students = present_class_a.intersection(present_class_b) 

# 2.Difference 
different_in_a = present_class_a - present_class_b) 


different_in_b = present_class_b - present_class_a 

# 3.Union
all_present_students =  present_class_a.union(present_class_b) 

# 4.Symmetric Difference 
present_in_exactlyOne = present_class_a ^ present_class_b 
```

```python
#displaying the attendance analysis 

all_students = set(students)

absent_students = all_students - all_present_students

print("======================= Attendance Analysis ==========================" , end = "\n\n") 

print("Total Unique Students : " , len(all_students) , end = "\n\n") 
print("All the Registered Students : \n" , all_students) 
print("Total  Present Students : " , len(all_present_students)) 
print("All the Present Students : \n" , all_present_students) 
print("Total Absent Student : ", len(absent_students))

print() 

print("Students Present in class A : " , len(present_class_a) )
print("Students Present in class B :  " , len(present_class_b) ) 

print() 

print("Students whose attendance in marked in both classes : \n" , common_students)
print("Students in Class A not in B : \n" , different_in_a )
print("Students in Class A not in B : \n" , different_in_b )
print("Students present in Exactly one Class : \n" , present_in_exactlyOne ) 




```

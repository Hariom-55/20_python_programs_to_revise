```python
import sys
probabilities = [] 

while True:
    try :
        probability = float(input("Enter the Probabilities: "))

        if  0 <= probability <=1  :
            probabilities.append(probability) 

        else :
            raise ValueError

    except ValueError :
        print("Invalid Input for Probability")
        continue 

    except EOFError :
        break


total_predictions = len(probabilities) 

if total_predictions > 0 :
    highest_probability = probabilities[0]
    lowest_probability = probabilities[0] 
    sum_of_probabilities = 0
    
    for probability in probabilities:
        sum_of_probabilities += probability 
        
        if probability > highest_probability :
            highest_probability = probability 
    
        elif probability < lowest_probability:
            lowest_probability = probability
    
    
    average_probability = sum_of_probabilities / total_predictions 
    
    #confidence classification 
    
    very_high = high = medium = low = 0   # initializing each type of confidence 
    
    for probability in probabilities :
        if  probability >= 0.90 :
            very_high += 1
    
        elif probability >=0.75 :
            high += 1
    
        elif probability >=0.50 :
            medium += 1  
    
        else:
            low +=1
    
    #finding duplicate predictions 
    duplicate_predictions = []
    
    for i in range ( total_predictions ) :
        for j in range( i+1, total_predictions ) : 
            if  probabilities[i] == probabilities[j] :
                if probabilities[i] not in duplicate_predictions :
                   duplicate_predictions.append(probabilities[i])
    
    high_confidence_probabilities  = [ prob  for prob in probabilities if prob >= 0.80 ]
    
    high_confidence_probabilities.sort( reverse = True ) 

else :
    sys.exit()

#printing the report 

print("============= Prediction Probability Report =============")
print( "Total Predictions: ", total_predictions ) 
print( "Recorded Predictions: " , probabilities ) 
print( "Highest Probability: " , highest_probability ) 
print( "Lowest Probability: " , lowest_probability ) 
print( "average Probability: " , average_probability ) 

print( "Confidence Classification of Predictions" )
print( f"""
Very High Confidence : {very_high}
High Confidence : {high}
Medium Confidence : {medium}
Low Confidence : {low}
""" ) 

print( "High Confidence Probabilities: " , high_confidence_probabilities ) 
print( "Duplicate Probabilities: " , duplicate_predictions)

probabilities.sort(reverse = True) # sorting the probabilities to rank them 
for index , value in enumerate (probabilities) :
    print(f"Rank {index+1} : {value} ")
```
